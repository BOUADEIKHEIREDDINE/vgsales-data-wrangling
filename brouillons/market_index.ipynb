### 1. Part de marché par genre

part de marché = Global_sales du genre / sum of global sales

```python
part_de_marche_genre = (df.groupby("Genre")["Global_Sales"].sum() / df["Global_Sales"].sum()) * 100
```

```python
part_de_marche_genre.sort_values(ascending = False, inplace = True)
```

```python
part_de_marche_genre
```

```python
part_de_marche_genre.plot(
    kind="pie",
    figsize=(10, 10),
    autopct="%1.1f%%",
    startangle=90,
    ylabel="",
    title="Part de marché par genre"
)
plt.axis("equal")
plt.show()
```

### Leadership (>30% d’un genre).

```python
leadership_genre = part_de_marche_genre[part_de_marche_genre > 30]
```
